## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

```python
methodology = """
Methodology:
- Baseline model: Random Forest classifier with balanced class weights.
- Features: measured page-level features such as impressions, average position,
  days since last update, word count, and CTR.
- Label: is_declining_label, defined from the observed decline proxy.
- Validation: compared a random row holdout with a client-grouped holdout
  to test performance on clients not used for training.
- Leakage checks: audited label-derived, overlapping-window, decision-derived,
  and ID fields. An intentional leakage attack was also used to confirm that
  the audit could detect label-derived information.
- The model output is used only as decision-support for human review.
"""

print(methodology)
```

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

```python
results = """
Results and findings:

The model produced a measured ranking of pages by probability of the
observed decline-label proxy.

The random row split gave a Precision@50 of about 0.90, while the
client-grouped holdout gave a more conservative Precision@50 of about 0.74.
This observed difference suggests that random row validation can give a
more optimistic estimate when pages from the same client appear in both
training and test data.

The grouped result should therefore be treated as the more relevant
directional measurement for performance on unseen clients.

These results support decision-support and prioritisation, not causal
claims or guaranteed future traffic outcomes.
"""

print(results)
```

## 5. Limitations

*What this work cannot claim.*

```python
limitations = """
Limitations:

1. The dataset is an anonymized starter dataset, so the findings may not
   represent every FlyRank client or every production environment.

2. The client-grouped validation result is based on a limited number of
   held-out clients, so it should be treated as directional rather than
   definitive.

3. The decline label is an observed proxy and does not prove that a page
   will decline in the future.

4. The model identifies associations useful for prioritisation; it does not
   establish that changing a page will cause traffic or ranking improvement.

5. The output should therefore be used as decision-support for human review,
   not as an automatic publishing or refresh decision.
"""

print(limitations)
```

*The action playbook output — the paper's recommendations section.*

```python
action_playbook = """
Action playbook:

1. Review the highest-ranked pages first using the model's measured decline
   probability.

2. Check the underlying page metrics and recent update history before taking
   any action.

3. Treat high-probability pages as candidates for investigation, not as
   guaranteed future declines.

4. Use human review to decide whether a content refresh or other intervention
   is appropriate.

5. Do not treat the model output as evidence that a refresh will recover
   traffic or as a prediction of Google's ranking algorithm.
"""

print(action_playbook)
```
